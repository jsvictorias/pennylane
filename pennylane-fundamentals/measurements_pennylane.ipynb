```python
import pennylane as qp
from pennylane import numpy as np 

qml = qp
```

```python
dev = qp.device('default.qubit', wires=1, shots=10)

@qp.qnode(dev)
def circuit():
    qp.H(wires=0)

    return qp.sample(qp.X(wires=0))

print(circuit())
```

```python
dev = qp.device('default.qubit', wires=1, shots=1000)

@qp.qnode(dev)
def circuit():
    qp.H(wires=0)

    return qp.counts()

print(circuit())
```

```python
dev = qp.device('default.qubit', wires=1, shots=1000)

@qp.qnode(dev)
def circuit():
    qp.H(wires=0)

    return qp.counts(qp.X(wires=0))

print(circuit())
```

```python
dev = qp.device('default.qubit', wires=2)

@qp.qnode(dev)
def circuit():
    qp.X(wires=1)

    return qp.expval(qp.Z(0))

circuit()
```

```python
dev = qp.device('default.qubit', wires=2)

@qp.qnode(dev)
def circuit():
    qp.RY(np.pi/4, wires = 0)
    qp.RX(np.pi/3, wires = 1)
    qp.CNOT(wires = [0, 1])

    return qp.expval(1/3*qp.PauliZ(0) @ qp.PauliZ(1))

circuit()
```
