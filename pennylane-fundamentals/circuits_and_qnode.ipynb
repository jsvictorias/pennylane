```python
@qp.qnode(dev)
def my_first_quantum_function(theta):
    qp.RX(theta, wires=0)
    qp.PauliY(wires=1)
    qp.Hadamard(wires=0)
    qp.Hadamard(wires=1)

    return qp.state()

print(my_first_quantum_function(np.pi/4))
```

```python
def my_first_quantum_function(theta):
    qp.RX(theta, wires=0)
    qp.PauliY(wires=1)
    qp.Hadamard(wires=0)
    qp.Hadamard(wires=1)

    return qp.state()

other_qnode_way = qp.QNode(my_first_quantum_function, dev)
print(other_qnode_way(np.pi/4))
```

```python
dev = qp.device("default.qubit", wires=["target", "control"])

@qp.qnode(dev)
def entangled():
    qp.Hadamard(wires='control')
    qp.CNOT(wires=['control', 'target'])

    return qp.state()

print(entangled)
```

### Subcircuits

```python
def subcircuit1(angle):
    qp.RX(angle, wires = 0)
    qp.PauliY(wires=1)
```

```python
def subcircuit2():
    qp.Hadamard(wires=0)
    qp.CNOT(wires=[0,1])
```

```python
def full_circuit(theta, phi):
    subcircuit1(theta)
    subcircuit2()
    subcircuit1(phi)
```

```python
theta = 0.3
phi = 0.2
print(qp.draw(full_circuit)(theta, phi))
```
