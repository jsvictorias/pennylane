```python
import pennylane as qp 
from pennylane import numpy as np 

qml = qp
```

```python
dev = qp.device('default.qubit', wires=2)

@qp.qnode(dev)
def circuit(state=None):
    qp.StatePrep(state, wires=range(2))
    return qp.state()

state = circuit([1/2,1/2,1/2,1/2])
print(qp.draw(circuit)(state))
```

### Basis State

```python
dev = qp.device("default.qubit", wires=2)

@qp.qnode(dev)
def circuit_2():
    qp.BasisChange(np.array([1,0,0]), wires=range(2))
    return qp.state()

print(circuit_2)
```

```python
dev = qp.device('default.qubit', wires=2)

@qp.qnode(dev)
def qubit_unitaries(U,V):
    qp.QubitUnitary(U, wires=0)
    qp.QubitUnitary(V, wires=[0,1])
    return qp.state()

print(qubit_unitaries(U,V))
```
